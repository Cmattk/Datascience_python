# K means Clustering 
K-Means Clustering is an Unsupervised Machine Learning algorithm which groups the unlabeled dataset into different clusters.

# Steps in K-Means Algorithm
Choose the number of clusters (K).

Randomly initialize K cluster centroids.

Assign each data point to the nearest cluster centroid.

Compute the new centroids by taking the mean of all points in each cluster.

Repeat steps 3-4 until centroids no longer change (or max iterations reached).

# Implementation of K-Means Clustering in Python

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
# Generate synthetic dataset
X, y = make_blobs(n_samples=300, centers=4, cluster_std=0.6, random_state=42)

# Visualize the dataset
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title("Generated Data for Clustering")
plt.show()
```

# Apply K-Means Clustering

```python
# Define the KMeans model
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

# Fit the model to the data
kmeans.fit(X)

# Get the cluster labels and centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_
```

```python
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', s=50)
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=200, label="Centroids")
plt.title("K-Means Clustering Results")
plt.legend()
plt.show()
```

# Choosing the Right Number of Clusters (Elbow Method)

```python
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42, n_init=10)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)  # Inertia is the sum of squared distances to centroids

# Plot the elbow curve
plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method for Optimal K")
plt.show()
```

the "elbow" point where the decrease in WCSS slows down—this suggests the best value of K.

# Project: Customer Segmentation Using K-Means

This project segments customers based on their Annual Income and Spending Score to help businesses target different groups effectively.

The dataset available on Kaggle: Mall_Customers.csv. or in the dataset file

```python
import pandas as pd  # For data handling
import numpy as np  # For numerical operations
import matplotlib.pyplot as plt  # For visualization
import seaborn as sns  # For advanced visualizations
from sklearn.cluster import KMeans  # K-Means clustering algorithm
from sklearn.preprocessing import StandardScaler  # For feature scaling (optional)

import warnings

warnings.filterwarnings('ignore')
```

```python
# Load dataset
df = pd.read_csv("C:\\Users\\franc\\Datascience_python\\dataset\\Mall_Customers.csv")

# Display first few rows
df.head()
```

```python
# Select Relevant Features, cluster based on spending habits
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]

# Annual Income (k$): Yearly income in thousands of dollars.
# Spending Score (1-100): Score assigned by the mall (higher means more spending).

```

```python
# Finding the Optimal Number of Clusters (Elbow Method)
# The Elbow Method helps determine the best K value by plotting the WCSS (Within-Cluster Sum of Squares).
# WCSS measures how close data points are to their cluster centroids.

# Initialize an empty list to store WCSS values
wcss = []

# Compute WCSS for different values of K (from 1 to 10)
for i in range(1,11):
    kmeans = KMeans(n_clusters=i, random_state=42, n_init=10) # Create KMeans model
    kmeans.fit(X) # Fit model to data
    wcss.append(kmeans.inertia_) # Store inertia (WCSS)

# plot the elbow curve
plt.figure(figsize=(8,5))
plt.plot(range(1,11), wcss, marker ='o', linestyle = '--', color = 'b')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS (Within-Cluster Sum of Squares)")
plt.title("Elbow Method for Optimal K")
plt.show()
```

```python
#  Applying K-Means Clustering
#  assume the best K = 5.
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)

# Fit the model to data
df['Cluster'] = kmeans.fit_predict(X)  # Assign each customer to a cluster

# Get cluster centers
centroids = kmeans.cluster_centers_

# Print cluster centers
print("Cluster Centers:\n", centroids)
```

```python
plt.figure(figsize=(10, 6))

# Scatter plot of customers, color-coded by cluster
sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)',
                hue='Cluster', palette='viridis', s=100)

# Plot centroids
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=300, label="Centroids")

plt.title("Customer Segmentation using K-Means")
plt.legend()
plt.show()
```
