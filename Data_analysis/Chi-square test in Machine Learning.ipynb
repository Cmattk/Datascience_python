# Chi-Square test:
helps us determine if there is a significant relationship between two categorical variables and the target variable. It is a non-parametric statistical test meaning it doesn’t follow normal distribution.

It checks whether there’s a significant difference between expected and observed frequencies, guiding us in identifying which features should be included in our model for better predictions.

Xc,^2 = sum of (O i,j - E i,j)^2 / E i,j  
c - degre of freedom.
O i,j - the observed frequency in cell ij.
E i,j is the expected frequency in cell ij.

Observed Values (O): Observed values refer to the actual data collected during an experiment or survey. These values represent real-world counts of occurrences within different categories.

Expected Values (E): Expected values represent the counts we would expect to see if there were no relationship between the variables. These values are calculated based on the assumption that the two variables are independent.

Contingency Table: A contingency table also called a cross-tabulation or two-way table is a structured way to display the distribution of two categorical variables. It helps in organizing observed frequencies before performing statistical analysis

Degrees of Freedom (df): It tell us the number of independent values that can vary when calculating statistics. It helps in determining the critical value from the Chi-Square distribution table.

# Types of Chi-Square test
The two main types are the chi-square test for independence and the chi-square goodness-of-fit test.

***Chi-Square Test for Independence:*** 
This test is used test wethere there is a significant relationship between two categorical variables.

It is used to determine whether changes in one variable are independent of changes in another. 
This test is applied when we have the counts of values for two nominal or categorical variables.
To conduct this test two requirements must be met: 
independence of observations and a relatively large sample size.

We test if shopping preference (Electronics, Clothing, Books) is related to payment method (Credit Card, Debit Card, PayPal). 
The null hypothesis assumes no relationship between them.

***Chi-Square Goodness-of-Fit Test:***
The Chi-Square Goodness-of-Fit test is used in statistical hypothesis testing to ascertain whether a variable is likely from a given distribution or not.

This test can be applied in situations when we have value counts for categorical variables. With the help of this test we can determine whether the data values are a representative sample of the entire population or if they fit our hypothesis well.

Imagine you are testing the fairness of a six-sided die.
The null hypothesis is that each face of the die should have an equal probability of landing face up.
In other words the die is unbiased and the proportions of each number (1 through 6) occurring are expected to be equal.

# steps to perform Chi-Square test
1: Define Your Hypotheses: Every statistical test starts with a hypothesis.

Null Hypothesis (H₀): The two variables are independent (no relationship).
Alternative Hypothesis (H₁): The two variables are related (there is a relationship).

2: Create a Contingency Table: This is simply a table that displays the frequency distribution of the two categorical variables.

3: Calculate Expected Values: To find the expected value for each cell use this formula:

E ij = (Row Total × Column Total) / Grand Total

4: Compute the Chi-Square Statistic: Now use the Chi-Square formula:

x2 = ∑ (O ij​–E ij​)2 / E ij

where:
O₍ᵢⱼ₎ = Observed value
E₍ᵢⱼ₎ = Expected value

If the observed and expected values are very different the Chi-Square value will be high which indicate a strong relationship.

5: Compare with the Critical Value: To determine if the result is significant we check:

If χ2 is greater than the critical value → Reject H₀ (There is a relationship).
If χ2 is smaller than the critical value → Fail to reject H₀ (No relationship).


# The Chi-Square Test helps us to find relationships or differences between categories. Here are the main uses of the Chi-Square Test:

1. Feature Selection in Machine Learning:
In machine learning the Chi-Square Test helps us to decide whether a categorical
feature e.g., color, product type is important for predicting the target variable e.g., sales, customer satisfaction.

It is especially useful when we are choosing features for classification tasks e.g., for models like decision trees or Naive Bayes to improve model performance.

2. Testing Independence:
The Chi-Square Test helps to check if two categorical variables are independent or related.
For example you might use it to see if customer demographics like age, gender are related to their product preferences and to understand patterns in your data.

3. Assessing Model Fit:
Sometimes the Chi-Square Test is used to see if a model’s predictions match the actual data. This is particularly helpful when working with classification models.
It checks whether the model is correctly predicting categorical outcomes and helping you to improve its accuracy.


We calculate Chi-square between each feature and the target and select the desired number of features with best Chi-square scores. 
Features that show significant dependencies with the target variable are considered important for prediction and can be selected for further analysis.

Feature selection is also known as **attribute selection** is a process of extracting the most relevant features from the dataset and then applying machine learning algorithms for the better performance of the model. 
A large number of irrelevant features increases the training time exponentially and increase the risk of overfitting.

# example 
Let’s examine a dataset with features including “income level” (low, medium, high) and “subscription status” (subscribed, not subscribed) indicate whether a customer subscribed to a service. 
The goal is to determine if this feature is relevant for predicting subscription status.

## 1: Make Hypothesis

Null hypothesis: No significant association between features.
Alternate Hypothesis: There is a significant association between features.


## 2: Contingency table


                Subscribed    Not subscribed    Row Total
Low Income         20           30               50 
Medium Income      40           25               65 
High Income        10           15               25 
Column Total       70           70               140 

## 3: Now calculate the expected frequencies:
e.g expected frequency for “Low Income” and “Subscribed” would be:

As Total count for each row Ri is 70 and. 
Total number of observations are 140.
Low Income, subscribed = (50×70)÷140=25

## expected frequency:

             Subscribed    Not subscribed 
Low Income         25           25                
Medium Income      32.5         32.5          
High Income        12.5         12.5               
               
## 4: Calculate the Chi-Square Statistic:

              Subscribed(O)    Not subscribed(O)    subcribed(E)
Low Income         20           30                     25               
Medium Income      40           25                     32.5
High Income        10           15                     12.5

χ2 =  (20-25)^2 / 25 + (40-32.5)^2 / 32.5 + (10-12.5)^2 / 12.5
X^2 = 3.231

5: Degrees of Freedom

Degrees of Freedom (df) = (3-1)  * (2-1) = 2.

## 6: Interpretations

Now you can compare the calculated value χ2 (3.231) with the critical value from the Chi-Square distribution table or any statistical software tool with 2 degrees of freedom. 
If the χ2 value is greater than the critical value you would reject the null hypothesis. 
This suggests that there is a significant association between “income level” and “subscription status,” and “income level” is a relevant feature for predicting subscription status. 
Let’s find out using python’s scipy library.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
df = 2
alpha = 0.05

critical_value = stats.chi2.ppf(1 - alpha, df)
critical_value
```

# a step-by-step breakdown:

1. `df = 2` - Sets the degrees of freedom for the chi-square distribution

2. `alpha = 0.05` - Sets the significance level (commonly used 5%)

3. `stats.chi2.ppf(1 - alpha, df)` - Uses the percent point function (PPF) from scipy.stats to find the critical value:
   - `1 - alpha` calculates the right-tail probability (0.95 in this case)
   - `df` specifies the degrees of freedom (2 in this case)

This code is typically used in hypothesis testing to determine the rejection region for a chi-square test.

critical value for df = 2 and α= 0.05 is 5.991. 
Since critical value is greater than chi-square value we accept the null hypothesis. 

Let’s visualize the chi-square distribution and Critical Region with python code.

```python
calculated_chi_square = 3.231

x = np.linspace(0, 10, 1000)
y = stats.chi2.pdf(x, df)

plt.plot(x, y, label='Chi-Square Distribution (df=2)')
plt.fill_between(x, y, where=(x > critical_value), color='red', alpha=0.5, label='Critical Region')
plt.axvline(calculated_chi_square, color='blue', linestyle='dashed', label='Calculated Chi-Square')
plt.axvline(critical_value, color='green', linestyle='dashed', label='Critical Value')
plt.title('Chi-Square Distribution and Critical Region')
plt.xlabel('Chi-Square Value')
plt.ylabel('Probability Density Function')
plt.legend()
```

# The plotting code step by step:

1. `x = np.linspace(0, 10, 1000)` - Creates an array of 1000 evenly spaced points between 0 and 10
2. `y = stats.chi2.pdf(x, df)` - Calculates the probability density function of chi-square distribution

3. Plot elements:
- `plt.plot(x, y, label='Chi-Square Distribution (df=2)')` - Plots the chi-square distribution curve
- `plt.fill_between()` - Fills the area above the critical value in red to show the critical region
- `plt.axvline(calculated_chi_square)` - Adds a blue dashed line for the calculated chi-square value
- `plt.axvline(critical_value)` - Adds a green dashed line for the critical value

4. Plot formatting:
- `plt.title()` - Sets the plot title
- `plt.xlabel()` - Labels the x-axis
- `plt.ylabel()` - Labels the y-axis
- `plt.legend()` - Adds a legend showing what each element represents

This code creates a visualization of a chi-square distribution with marked critical regions and test statistics.

The green dashed line represents the critical value the threshold beyond which you would reject the null hypothesis.

The red dashed line represents the critical value (5.991) for a significance level of 0.05 with 2 degrees of freedom.

The shaded area to the right of the critical value represents the rejection region.

If the calculated Chi-Square statistic falls within this shaded area then you would reject the null hypothesis. 
The calculated chi-square value does not fall within the critical region therefore accepting the null hypothesis. 
Hence there is no significant association between two variables.
