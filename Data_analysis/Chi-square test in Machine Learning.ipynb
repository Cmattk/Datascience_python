Chi-Square test helps us determine if there is a significant relationship between two categorical variables and the target variable. It is a non-parametric statistical test meaning it doesn’t follow normal distribution.

It checks whether there’s a significant difference between expected and observed frequencies, guiding us in identifying which features should be included in our model for better predictions.

Xc,^2 = sum of (O i,j - E i,j)^2 / E i,j  
c - degre of freedom.
O i,j - the observed frequency in cell ij.
E i,j is the expected frequency in cell ij.

Observed Values (O): Observed values refer to the actual data collected during an experiment or survey. These values represent real-world counts of occurrences within different categories.

Expected Values (E): Expected values represent the counts we would expect to see if there were no relationship between the variables. These values are calculated based on the assumption that the two variables are independent.

Contingency Table: A contingency table also called a cross-tabulation or two-way table is a structured way to display the distribution of two categorical variables. It helps in organizing observed frequencies before performing statistical analysis

Degrees of Freedom (df): It tell us the number of independent values that can vary when calculating statistics. It helps in determining the critical value from the Chi-Square distribution table.

Types of Chi-Square test
The two main types are the chi-square test for independence and the chi-square goodness-of-fit test.

***Chi-Square Test for Independence:*** 
This test is used test wethere there is a significant relationship between two categorical variables.

It is used to determine whether changes in one variable are independent of changes in another. 
This test is applied when we have the counts of values for two nominal or categorical variables.
To conduct this test two requirements must be met: 
independence of observations and a relatively large sample size.

We test if shopping preference (Electronics, Clothing, Books) is related to payment method (Credit Card, Debit Card, PayPal). 
The null hypothesis assumes no relationship between them.

***Chi-Square Goodness-of-Fit Test:***
The Chi-Square Goodness-of-Fit test is used in statistical hypothesis testing to ascertain whether a variable is likely from a given distribution or not.

This test can be applied in situations when we have value counts for categorical variables. With the help of this test we can determine whether the data values are a representative sample of the entire population or if they fit our hypothesis well.

Imagine you are testing the fairness of a six-sided die.
The null hypothesis is that each face of the die should have an equal probability of landing face up.
In other words the die is unbiased and the proportions of each number (1 through 6) occurring are expected to be equal.

**steps to perform Chi-Square test**
1: Define Your Hypotheses: Every statistical test starts with a hypothesis.

Null Hypothesis (H₀): The two variables are independent (no relationship).
Alternative Hypothesis (H₁): The two variables are related (there is a relationship).

2: Create a Contingency Table: This is simply a table that displays the frequency distribution of the two categorical variables.

3: Calculate Expected Values: To find the expected value for each cell use this formula:

E ij = (Row Total × Column Total) / Grand Total

4: Compute the Chi-Square Statistic: Now use the Chi-Square formula:

x2 = ∑ (O ij​–E ij​)2 / E ij

where:
O₍ᵢⱼ₎ = Observed value
E₍ᵢⱼ₎ = Expected value

If the observed and expected values are very different the Chi-Square value will be high which indicate a strong relationship.

5: Compare with the Critical Value: To determine if the result is significant we check:

If χ2 is greater than the critical value → Reject H₀ (There is a relationship).
If χ2 is smaller than the critical value → Fail to reject H₀ (No relationship).


The Chi-Square Test helps us to find relationships or differences between categories. Here are the main uses of the Chi-Square Test:

1. Feature Selection in Machine Learning:
In machine learning the Chi-Square Test helps us to decide whether a categorical
feature e.g., color, product type is important for predicting the target variable e.g., sales, customer satisfaction.

It is especially useful when we are choosing features for classification tasks e.g., for models like decision trees or Naive Bayes to improve model performance.

2. Testing Independence:
The Chi-Square Test helps to check if two categorical variables are independent or related.
For example you might use it to see if customer demographics like age, gender are related to their product preferences and to understand patterns in your data.

3. Assessing Model Fit:
Sometimes the Chi-Square Test is used to see if a model’s predictions match the actual data. This is particularly helpful when working with classification models.
It checks whether the model is correctly predicting categorical outcomes and helping you to improve its accuracy.
