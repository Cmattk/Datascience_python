# Which Database to Choose in System Design

## 🗂️ Key Features of SQL Databases

| Feature                             | Description                                                                                            |
| ----------------------------------- | ------------------------------------------------------------------------------------------------------ |
| **Tabular Data Model**              | Organizes data in tables with rows (records) and columns (attributes).                                 |
| **Fixed Schema**                    | Requires predefined schema before inserting data; ideal for structured data.                           |
| **ACID Compliance**                 | Ensures **Atomicity**, **Consistency**, **Isolation**, and **Durability**—critical for data integrity. |
| **Structured Query Language (SQL)** | Standardized and powerful query language for complex queries and reporting.                            |
| **Strong Relationships**            | Supports foreign keys and joins; excels at managing complex data relationships.                        |

### 🔧 Common SQL Databases

* **MySQL** – Open-source, widely used.
* **PostgreSQL** – Advanced features, extensibility, open-source.

---

## 🔄 What is a NoSQL Database?

| Feature                        | Description                                                                                   |
| ------------------------------ | --------------------------------------------------------------------------------------------- |
| **Flexible Data Model**        | Supports multiple data models: key-value, document, wide-column, graph.                       |
| **Schema-less**                | No predefined schema required; data can evolve dynamically.                                   |
| **BASE Model**                 | Prioritizes **Basically Available**, **Soft State**, and **Eventually Consistent** over ACID. |
| **Proprietary Query Language** | Each NoSQL type may have a unique query language optimized for its structure.                 |


### 🔧 Common NoSQL Databases

* **MongoDB** – Document-oriented, flexible and scalable.
* **Cassandra** – Wide-column store, optimized for write-heavy applications.

---

## ⚖️ SQL vs NoSQL – Full Comparison Table

| Category           | SQL                                                  | NoSQL                                                           |
| ------------------ | ---------------------------------------------------- | --------------------------------------------------------------- |
| **Data Model**     | Tabular (relational)                                 | Key-Value, Document, Column-Family, Graph                       |
| **Schema**         | Rigid, predefined                                    | Schema-less, dynamic                                            |
| **Scalability**    | Vertical (scale up)                                  | Horizontal (scale out)                                          |
| **Transactions**   | ACID                                                 | BASE (Eventually consistent)                                    |
| **Query Language** | Standard SQL                                         | Varies; some use custom languages or JSON-like queries          |
| **Relationships**  | Built-in via joins and constraints                   | Requires denormalization or manual handling                     |
| **Use Cases**      | Structured data, complex queries, strong consistency | Large-scale data, unstructured/semi-structured, flexible schema |
| **Examples**       | MySQL, PostgreSQL, Oracle, SQL Server                | MongoDB, Cassandra, Couchbase, Neo4j                            |

---

## 🔍 When to Use SQL

| Situation                      | Why SQL is Better                                                      |
| ------------------------------ | ---------------------------------------------------------------------- |
| **Complex Queries Needed**     | SQL supports joins, aggregations, subqueries for detailed analytics.   |
| **Data Integrity is Critical** | ACID compliance ensures consistency—ideal for banking, inventory, etc. |
| **Transactional Systems**      | Handles multi-step transactions reliably.                              |
| **Fixed, Structured Data**     | Works best when data structure is stable and well-defined.             |

### ❗ Challenges of SQL

* Hard to scale horizontally
* Schema changes require downtime or migrations
* Not ideal for large-scale distributed systems

---

## 🔍 When to Use NoSQL

| Situation                                | Why NoSQL is Better                                        |
| ---------------------------------------- | ---------------------------------------------------------- |
| **High Traffic / Big Data**              | Scales horizontally to handle massive loads.               |
| **Evolving Schema**                      | No need for schema migrations; supports rapid development. |
| **Real-Time Analytics / Streaming Data** | Fast ingest and flexible processing.                       |
| **Unstructured or Semi-Structured Data** | Easily stores JSON, key-value pairs, etc.                  |

### ❗ Challenges of NoSQL

* Less mature transaction support (limited ACID)
* Inconsistent data in eventual consistency models
* Complex queries are harder to write or inefficient
* Learning curve for custom query languages

---

## 🧠 Final Thoughts

### SQL: Choose when you need:

* Relational data
* Reliable transactions
* Strong data consistency
* Standardized querying

### NoSQL: Choose when you need:

* Horizontal scalability
* Flexible, evolving schema
* Fast writes and reads at scale
* High availability over strong consistency

# File and Database Storage Systems in System Design

File and database storage systems are important to the effective management and arrangement of data in system design. These systems offer a structure for data organization, retrieval, and storage in applications while guaranteeing data accessibility and integrity. Database systems provide structured data management with advanced querying capabilities, whereas file systems manage both structured and unstructured data.

---

## File-Based Storage System

Stores data as individual files in a hierarchical directory system (folders and subfolders). Common formats: `.txt`, `.json`, `.csv`, `.log`, images, videos, etc.

### Pros:

| Advantage          | Description                                              |
| ------------------ | -------------------------------------------------------- |
| **Simplicity**     | Easy to implement; no need for database setup.           |
| **Compatibility**  | Universally supported by operating systems and software. |
| **Cost-Effective** | No licensing or specialized infrastructure required.     |
| **Flexibility**    | Supports any file type (e.g., documents, media).         |
| **Low Overhead**   | Ideal for static or infrequently accessed data.          |

### Cons:

| Disadvantage                    | Description                                              |
| ------------------------------- | -------------------------------------------------------- |
| **Limited Scalability**         | Not suitable for large, complex systems.                 |
| **No Query Support**            | Cannot perform advanced or relational queries.           |
| **No Data Integrity Mechanism** | Difficult to maintain consistency between related files. |
| **Duplication & Redundancy**    | Hard to enforce uniqueness or prevent duplication.       |

### 🔧 Use Cases:

* Log storage
* Image repositories
* Static content (e.g., PDFs, configuration files)
* Simple backup systems

---

## 🗄️ **Database Storage System**

Stores data in a structured form (tables, rows, columns). Supports indexing, querying, and relationships between data entities.

### Pros:

| Advantage                     | Description                                                                 |
| ----------------------------- | --------------------------------------------------------------------------- |
| **Efficient Querying**        | Supports powerful querying (e.g., SQL) for data retrieval and manipulation. |
| **Data Relationships**        | Enables normalization, foreign keys, joins, etc.                            |
| **Data Integrity**            | Enforces constraints and transaction guarantees (e.g., ACID).               |
| **Scalability**               | Scales vertically or horizontally (e.g., via sharding/replication).         |
| **Security & Access Control** | Role-based permissions and auditing features.                               |

### Cons:

| Disadvantage             | Description                                                      |
| ------------------------ | ---------------------------------------------------------------- |
| **Complex Setup**        | Requires schema design, deployment, and configuration.           |
| **Higher Cost**          | May involve licensing, infrastructure, and management overhead.  |
| **Performance Overhead** | May be overkill for basic storage tasks (e.g., flat file reads). |

### 🔧 Use Cases:

* Transactional systems (e.g., e-commerce, banking)
* Relational data modeling (e.g., HR, inventory)
* Real-time analytics
* Applications requiring multi-entity consistency

---

## ⚖️ **Comparison Table: File vs Database Storage**

| Feature                      | File-Based Storage System  | Database Storage System                       |
| ---------------------------- | -------------------------- | --------------------------------------------- |
| **Structure**                | Flat files in folders      | Tables with rows and columns                  |
| **Schema**                   | None required              | Fixed (SQL) or flexible (NoSQL) schema        |
| **Querying**                 | Manual or basic read/write | SQL or other query languages                  |
| **Data Relationships**       | Not supported              | Fully supported via constraints & joins       |
| **Integrity & Validation**   | Minimal, manual            | Enforced via schema, constraints, ACID        |
| **Scalability**              | Limited                    | Scales horizontally or vertically             |
| **Security**                 | OS-level permissions       | Fine-grained access control                   |
| **Cost**                     | Low or none                | Medium to high, depending on DBMS             |
| **Performance (Simple Ops)** | High                       | May have slight overhead                      |
| **Use Case Fit**             | Documents, media, logs     | Structured records, business logic, analytics |

---

## 🥇 **When to Use Which?**

### ✅ Choose **File-Based Storage** When:

* You need to store large media or logs.
* You're building a lightweight or temporary system.
* You want to avoid the overhead of database management.
* You're storing unstructured or binary data.

### Choose **Database Storage** When:

* You require advanced data querying and filtering.
* Data relationships and referential integrity are essential.
* You need transactional support (e.g., ACID).
* You're working with structured or semi-structured data.
* Your system must scale with growing data complexity.

---

## Summary

| You Need...                                    | Use...             |
| ---------------------------------------------- | ------------------ |
| High performance for simple file access        | File-Based Storage |
| Flexible schema & fast prototyping             | NoSQL Database     |
| Complex relationships & strict data validation | SQL Database       |
| Durability and auditability                    | Database           |
| Quick and cheap setup                          | File System        |

# Database Replication in System Design

Database replication is essential to system design, particularly when it comes to guaranteeing data scalability, availability, and reliability. It involves building and keeping several copies of a database on various servers to improve fault tolerance and performance.


Here's a **comprehensive and structured summary** of your detailed notes on **Database Replication**, formatted for clarity and future use (e.g., for study, documentation, or presentation purposes):

---

## 📌 **Importance of Database Replication**

| Benefit                  | Description                                                                      |
| ------------------------ | -------------------------------------------------------------------------------- |
| ✅ **High Availability**  | Ensures applications continue running even if one or more database servers fail. |
| 🔁 **Disaster Recovery** | Enables quick recovery by maintaining backups across multiple locations.         |
| ⚖️ **Load Balancing**    | Distributes read queries across replicas to improve performance.                 |
| 🛡️ **Fault Tolerance**  | Another replica can seamlessly take over in the event of failure.                |
| 🚀 **Scalability**       | Can scale read and write operations across multiple nodes.                       |
| 🌍 **Data Locality**     | Places data closer to users, reducing latency and improving UX.                  |

---

## ⚙️ **How Database Replication Works (Step-by-Step)**

1. **Primary Database Identified**: A master/source is chosen for all original writes.
2. **Replicas Set Up**: One or more secondary/target databases are connected.
3. **Change Capture**: Modifications in the primary DB are tracked (e.g., using binlogs or CDC).
4. **Transmit Changes**: Changes are sent to replicas over the network.
5. **Apply Changes**: Replicas update their data to reflect the primary’s state.
6. **Monitor Synchronization**: Ensure replicas remain consistent with the primary.
7. **Read/Write Operations**: Reads can be offloaded to replicas; writes depend on the replication model.

---

## 🔁 **Types of Database Replication**

| Type                             | Description                                                            | Use Case                           |
| -------------------------------- | ---------------------------------------------------------------------- | ---------------------------------- |
| **Master-Slave**                 | One master handles writes; replicas handle reads.                      | Read-heavy systems.                |
| **Master-Master (Multi-Master)** | Multiple masters accept writes; changes are synchronized.              | High availability + write scaling. |
| **Snapshot**                     | Periodic full copy of the database state.                              | Static or slow-changing data.      |
| **Transactional**                | Real-time replication of individual transactions.                      | Time-sensitive, OLTP workloads.    |
| **Merge**                        | Both publisher and subscribers can update; conflict resolution needed. | Disconnected or mobile clients.    |

---

## 🧩 **Replication Strategies**

| Strategy                  | Description                                          | Benefits                                     |
| ------------------------- | ---------------------------------------------------- | -------------------------------------------- |
| **Full Replication**      | Entire DB copied to all replicas.                    | Simplifies reads; good for smaller datasets. |
| **Partial Replication**   | Only specific tables/rows/columns are replicated.    | Reduces bandwidth/storage costs.             |
| **Selective Replication** | Conditional replication based on criteria.           | Fine-grained control.                        |
| **Sharding**              | Splits DB horizontally across servers.               | High write scalability.                      |
| **Hybrid**                | Mix of strategies for different parts of the system. | Customizable and adaptable.                  |

---

## 🔧 **Replication Configurations in System Design**

| Configuration        | Description                                          | Trade-Off                            |
| -------------------- | ---------------------------------------------------- | ------------------------------------ |
| **Synchronous**      | Waits for at least one replica to confirm the write. | High consistency, lower performance. |
| **Asynchronous**     | Primary doesn’t wait for replica acknowledgment.     | High performance, possible lag.      |
| **Semi-Synchronous** | A hybrid; waits for one replica, others async.       | Balanced consistency and speed.      |

---

## ⚠️ **Challenges with Database Replication**

| Challenge               | Explanation                                                             |
| ----------------------- | ----------------------------------------------------------------------- |
| **Data Consistency**    | Asynchronous replication may cause temporary mismatches.                |
| **System Complexity**   | Requires setup, monitoring, conflict resolution logic.                  |
| **Cost**                | More servers, network resources, and maintenance overhead.              |
| **Conflict Resolution** | In master-master setups, simultaneous writes need reconciliation.       |
| **Latency**             | Synchronous writes may slow down applications due to round-trip delays. |

---

## 📌 Quick Reference Table

| Feature                 | Master-Slave     | Master-Master     | Snapshot           | Transactional | Merge                    |
| ----------------------- | ---------------- | ----------------- | ------------------ | ------------- | ------------------------ |
| **Write Source**        | Only Master      | Any Master        | Periodic           | Master        | Any node                 |
| **Real-Time**           | Yes (async)      | Yes               | No                 | Yes           | Depends                  |
| **Conflict Resolution** | No               | Required          | No                 | No            | Required                 |
| **Use Case**            | Read scalability | High availability | Archival/reporting | OLTP          | Mobile/disconnected apps |

---

## 📚 When to Use Database Replication

* ✅ **Use it when**:

  * Your system must stay online (e.g., financial apps).
  * Read scaling is necessary (e.g., search engines).
  * Disaster recovery is critical.
  * You operate globally and want localized data access.

* ⚠️ **Avoid or limit** when:

  * You have tight consistency requirements across all nodes (e.g., bank ledgers using master-master without strong conflict resolution).
  * Your system has heavy write contention and low tolerance for conflicts.

# Database Sharding - System Design

Database sharding is a technique for horizontal scaling of databases, where the data is split across multiple database instances, or shards, to improve performance and reduce the impact of large amounts of data on a single database.

Each chunk/partition is known as a "shard" and each shard has the same database schema as the original database.
We distribute the data in such a way that each row appears in exactly one shard.
It's a good mechanism to improve the scalability of an application. 

---

## 🔢 **Methods of Sharding**

### 1. 🔑 Key-Based Sharding (Hash-Based)

* **How it works**: Applies a hash function on a shard key (e.g., user ID) to determine the target shard.
* **Example**: `application_id % number_of_shards`
* **Pros**:

  * Predictable data distribution
  * Good for range queries (if keys are well-chosen)
* **Cons**:

  * Uneven data distribution if hash function or key is poor
  * Hard to scale dynamically (rehashing required)
  * Choosing a proper key is critical

---

### 2. 📏 Range-Based Sharding (Horizontal Sharding)

* **How it works**: Data is divided based on value ranges of a key (e.g., names A–P in one shard, Q–Z in another).
* **Pros**:

  * Simple to understand
  * Efficient for range queries
* **Cons**:

  * Uneven distribution likely if ranges aren’t well planned
  * Queries spanning multiple ranges are harder

---

### 3. 🧱 Vertical Sharding

* **How it works**: Split data by feature/function. For instance, user profiles, tweets, and followers in separate databases.
* **Pros**:

  * Improved query performance (each shard optimized for a purpose)
  * Clear separation of concerns

* **Cons**:

  * Hotspots may form
  * Schema changes require coordination across shards

---

### 4. 📚 Directory-Based Sharding

* **How it works**: Maintains a centralized lookup service to map shard keys to specific shards.
* **Pros**:

  * Highly flexible
  * Easier dynamic scaling
* **Cons**:

  * Single point of failure (central directory)
  * Added latency due to lookup step

---

## 🔧 **Ways to Optimize Sharding**

| Strategy                          | Benefit                                                                           |
| --------------------------------- | --------------------------------------------------------------------------------- |
| **Use Consistent Hashing**        | Smooth data distribution, avoids massive reshuffling when adding/removing shards. |
| **Choose a Balanced Shard Key**   | Prevents hot spots and ensures even workload distribution.                        |
| **Monitor & Rebalance Regularly** | Adapts to changes in data patterns.                                               |
| **Use Automation Tools**          | Reduces manual effort and human error in shard assignment and migration.          |

---

## 🔄 **Alternatives to Sharding**

| Technique            | Description                                                                            |
| -------------------- | -------------------------------------------------------------------------------------- |
| **Vertical Scaling** | Increase resources (CPU/RAM/disk) on a single server. Simple, but limited.             |
| **Replication**      | Multiple read replicas for scaling reads and availability. May face write bottlenecks. |
| **Partitioning**     | Split data within one server (not across servers).                                     |
| **Caching**          | Use systems like Redis or Memcached to offload reads.                                  |
| **CDNs**             | For read-heavy static data (images/videos), use CDNs to reduce database load.          |

---

## 🟩 **Advantages of Sharding**

| Benefit                     | Description                                 |
| --------------------------- | ------------------------------------------- |
| ⚡ **Performance**           | Spreads load, reduces response time.        |
| 📈 **Scalability**          | Add more shards as data grows.              |
| 💡 **Resource Utilization** | Avoids overloading one server.              |
| 🔒 **Fault Isolation**      | Failure in one shard doesn’t affect others. |
| 💰 **Cost Efficiency**      | Can use smaller, less expensive servers.    |

---

## 🟥 **Disadvantages of Sharding**

| Challenge                   | Description                                                |
| --------------------------- | ---------------------------------------------------------- |
| 🧩 **Complexity**           | More moving parts to manage (e.g., metadata, consistency). |
| 🔄 **Rebalancing**          | Redistributing shards when uneven is hard.                 |
| 🔗 **Cross-Shard Queries**  | Harder to implement joins and aggregates across shards.    |
| 🔍 **Operational Overhead** | More backups, replication, maintenance.                    |
| 📉 **Data Loss Risk**       | If one shard fails and lacks backup, its data is gone.     |

---

## 📊 Summary Comparison of Sharding Methods

| Sharding Type       | Distribution      | Query Type         | Complexity | Use Case                                  |
| ------------------- | ----------------- | ------------------ | ---------- | ----------------------------------------- |
| **Key-Based**       | Hash-based        | Random or range    | Moderate   | Uniform workload distribution             |
| **Range-Based**     | Value range       | Range queries      | Low–Medium | Ordered data, analytics                   |
| **Vertical**        | Columns/features  | Targeted queries   | High       | Modular apps (e.g., microservices)        |
| **Directory-Based** | Mapped via lookup | Depends on mapping | High       | Complex data relationships, dynamic scale |

# Normalization Process in DBMS

Database Normalization is any systematic process of organizing a database schema such that no data redundancy occurs and there is least or no anomaly while performing any update operation on data. 

It means dividing a large table into smaller pieces such that data redundancy should be eliminated.

The normalizing procedure depends on the functional dependencies among the attributes inside a table and uses several normal forms to guide the design process.

## Why Normalize a Database.

✅ Eliminate data redundancy – Avoid storing the same data in multiple places.

✅ Ensure data integrity and consistency – Changes in one place reflect across the system.

✅ Make updates, inserts, and deletes efficient – Reduces chances of anomalies.

✅ Organize data logically – Easier to understand and manage.

## The Normalization Process Begins With the Following:

**First Normal Form (1NF):** 
Ensures that each column contains only atomic values that cannot be divided, and each record is unique.

Example (Before 1NF):

| StudentID | Name  | Subjects           |
| --------- | ----- | ------------------ |
| 1         | Alice | Math, English      |
| 2         | Bob   | Physics, Chemistry |

Violation: "Subjects" contains multiple values.

After 1NF:

| StudentID | Name  | Subject   |
| --------- | ----- | --------- |
| 1         | Alice | Math      |
| 1         | Alice | English   |
| 2         | Bob   | Physics   |
| 2         | Bob   | Chemistry |


**Second Normal Form (2NF)** – Remove Partial Dependencies

Must be in 1NF.

Every non-prime attribute must be fully functionally dependent on the entire primary key, not just part of it.

Example (Before 2NF):

| StudentID | CourseID | StudentName | CourseName |
| --------- | -------- | ----------- | ---------- |
| 1         | C101     | Alice       | Math       |
| 2         | C102     | Bob         | Physics    |

Problem:
StudentName depends only on StudentID,
CourseName depends only on CourseID.

After 2NF:
Split into two tables to remove partial dependencies.

Student Table:

| StudentID | StudentName |
| --------- | ----------- |
| 1         | Alice       |
| 2         | Bob         |

Course Table:

| CourseID | CourseName |
| -------- | ---------- |
| C101     | Math       |
| C102     | Physics    |

Enrollment Table:

| StudentID | CourseID |
| --------- | -------- |
| 1         | C101     |
| 2         | C102     |


**Third Normal Form (3NF)** – Remove Transitive Dependencies
Must be in 2NF.

No transitive dependencies, i.e., non-prime attribute depends on another non-prime attribute.

Example (Before 3NF):

| EmployeeID | Name  | Department | DeptLocation |
| ---------- | ----- | ---------- | ------------ |
| 1          | Alice | IT         | 2nd Floor    |
| 2          | Bob   | HR         | 3rd Floor    |

Problem:
DeptLocation depends on Department, which is not a key → transitive dependency.

After 3NF:
Employee Table:

| EmployeeID | Name  | Department |
| ---------- | ----- | ---------- |
| 1          | Alice | IT         |
| 2          | Bob   | HR         |

Department Table:

| Department | DeptLocation |
| ---------- | ------------ |
| IT         | 2nd Floor    |
| HR         | 3rd Floor    |


**Boyce-Codd Normal Form (BCNF)** – Stronger 3NF

For every functional dependency (A → B), A must be a super key.
Useful when a table has overlapping candidate keys.
It removes anomalies caused by overlapping candidate keys.

(Before BCNF):
university course allocation table:

| Course | Instructor | Room   |
| ------ | ---------- | ------ |
| CS101  | Dr. Smith  | Room A |
| CS101  | Dr. Smith  | Room B |

Functional Dependencies:
Course → Instructor ✅ (each course has one instructor)
Room → Instructor ❌ (but each room can only have one instructor at a time)

both Course and Room can determine Instructor, but neither is a super key, so this violates BCNF.

(After BCNF):
Course Table:

| Course | Instructor |
| ------ | ---------- |
| CS101  | Dr. Smith  |

Room Table:

| Room   | Instructor |
| ------ | ---------- |
| Room A | Dr. Smith  |
| Room B | Dr. Smith  |

Now each table has only one dependency and all determinants are super keys.

**Fourth Normal Form (4NF)**

A table is in 4NF if:
It is in BCNF, and it has no multi-valued dependencies (MVDs).

A student can enroll in multiple courses and join multiple clubs.

| Student | Course  | Club     |
| ------- | ------- | -------- |
| Alice   | Math    | Chess    |
| Alice   | Math    | Robotics |
| Alice   | English | Chess    |
| Alice   | English | Robotics |


This causes a combinatorial explosion: 2 courses × 2 clubs = 4 rows.

Two independent MVDs:
Student →→ Course
Student →→ Club

(After 4NF):
Student-Course Table:

| Student | Course  |
| ------- | ------- |
| Alice   | Math    |
| Alice   | English |

Student-Club Table:

| Student | Club     |
| ------- | -------- |
| Alice   | Chess    |
| Alice   | Robotics |

Now both tables avoid unnecessary duplication and remove multi-valued dependencies.

**Fifth Normal Form (5NF)**
A table is in 5NF (also called Projection-Join Normal Form) if:

It is in 4NF and cannot be decomposed further without losing data or meaning.

It deals with join dependencies — extremely rare and complex.

(Before 5NF):

| Product | Supplier | Customer |
| ------- | -------- | -------- |
| A       | S1       | C1       |
| A       | S1       | C2       |
| A       | S2       | C1       |
| A       | S2       | C2       |


Product-Supplier:

| Product | Supplier |
| ------- | -------- |
| A       | S1       |
| A       | S2       |

Product-Customer:

| Product | Customer |
| ------- | -------- |
| A       | C1       |
| A       | C2       |

Supplier-Customer:

# Denormalization in Databases
Denormalization focuses on combining multiple tables to make queries execute quickly. It adds redundancies in the database though. 

This can help us avoid costly joins in a relational database. Note that denormalization does not mean 'reversing normalization' or 'not to normalize'. It is an optimization technique that is applied after normalization.

# Step-by-Step Example

🎓 Context: School Database

1. Students

| student_id  | name  |
| ----------- | ----- |
| 1           | Alice |
| 2           | Bob   |

2. Courses

| course_id  | title   | teacher_id  |
| ---------- | ------- | ----------- |
| C1         | Math    | T1          |
| C2         | Physics | T2          |

3. Teachers

| teacher_id  | name      |
| ----------- | --------- |
| T1          | Mr. Smith |
| T2          | Ms. Stone |

4. Enrollments

| student_id  | course_id  |
| ----------- | ---------- |
| 1           | C1         |
| 2           | C2         |

To get: "Alice's course and teacher name", we need:

Join Students → Enrollments → Courses → Teachers.

SQL JOIN Query:

SELECT 
    s.student_id,
    s.name AS student_name,
    c.course_id,
    c.title AS course_title,
    t.name AS teacher_name
FROM students AS s
JOIN enrollments AS e ON s.student_id = e.student_id
JOIN courses AS c ON e.course_id = c.course_id
JOIN teachers AS t ON c.teacher_id = t.teacher_id;


# Denormalized Version

One denormalized table:

| student\_id | student\_name | course\_id | course\_title | teacher\_name |
| ----------- | ------------- | ---------- | ------------- | ------------- |
| 1           | Alice         | C1         | Math          | Mr. Smith     |
| 2           | Bob           | C2         | Physics       | Ms. Stone     |

Now we can retrieve the data directly with no joins.


🔧 When to Use Denormalization?  

✅ Use denormalization when:

    You're building read-heavy systems (e.g., analytics dashboards).

    You need low-latency reads (e.g., mobile APIs).

    Join operations become a performance bottleneck.

    You cache frequently joined results to reduce load.

🚫 Avoid denormalization if:

    Data integrity and consistency are top priorities (e.g., financial systems).

    The write/update operations are more frequent than reads.


🔚 Final Thoughts
Normalization ensures clean, consistent, and storage-efficient data.

Denormalization is a deliberate tradeoff to gain speed by sacrificing some purity.

Real-world systems (especially at scale) often use a hybrid approach — normalized schema for writes, denormalized views/tables for reads.
