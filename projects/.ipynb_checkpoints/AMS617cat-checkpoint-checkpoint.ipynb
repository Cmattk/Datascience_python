# Convert categorical variables to numerical

```python
# Convert categorical variables to numerical
df_encoded = df.copy()

# Identify categorical columns
categorical_columns = df_encoded.select_dtypes(include=['object']).columns

# Convert 'yes'/'no' to 1/0
binary_columns = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea']
for col in binary_columns:
    df_encoded[col] = df_encoded[col].map({'yes': 1, 'no': 0})

# Convert 'furnishingstatus' to numerical encoding
df_encoded['furnishingstatus'] = df_encoded['furnishingstatus'].map({'furnished': 2, 'semi-furnished': 1, 'unfurnished': 0})

print(df_encoded.head(3))
```

# c) Formulate Five Objectives and Perform Hypothesis Testing

```python
# neccesary libraries
import numpy as np
from scipy import stats
```

## Objective 1: Does area affect price?

```python
# Define hypothesis
null_hypothesis = "Area has no effect on price."
alternate_hypothesis = "Area has an effect on price."
alpha = 0.05

# Perform Pearson correlation test
corr, p_value = stats.pearsonr(df_encoded['area'], df_encoded['price'])

# condition
if p_value <= alpha:  
    decision = "Reject"
    conclusion = "There is sufficient evidence to claim area has an effect on price."
else:
    decision = "Fail to reject"
    conclusion = "There is statistically significant evidence that area has no effect on price."

# results
print(f"Correlation: {corr:.2f}, P-value: {p_value}")
print(f"Decision: {decision} the null hypothesis at alpha={alpha}.")
print("Conclusion:", conclusion)
```

## Objective 2: Do houses near the main road have a higher price?
mainroad is a categorical variable (yes/no mapped to 1/0),

we compare the means of house prices for both groups.

```python
# Define hypothesis
null_hypothesis = "Main road has no effect on price."
alternate_hypothesis = "Main road has an effect on price."
alpha = 0.05

# Perform t-test
t_stat, p_value = stats.ttest_ind(df_encoded['mainroad'],df_encoded['price'])

# condition
if p_value <= alpha:  
    decision = "Reject"
    conclusion = "There is sufficient evidence to claim main road has an effect on price."
else:
    decision = "Fail to reject"
    conclusion = "There is statistically significant evidence that main road has no effect on price."

# results
print(f"T-statistic: {t_stat:.4f}, P-value: {p_value}")
print(f"Decision: {decision} the null hypothesis at alpha={alpha}.")
print("Conclusion:", conclusion)
```

## Objective 3: Does the number of bedrooms impact price?
I am using an Ordinary Least Squares (OLS) regression to check if bedrooms significantly affect price.

```python
# neccesary libraries
import statsmodels.api as sm
```

```python
# Define hypothesis
null_hypothesis = "The number of bedrooms has no significant impact on the price."
alternate_hypothesis = "The number of bedrooms has significant impact on the price."

# Define independent (X) and dependent (y) variables
X = df_encoded[['bedrooms']]
y = df_encoded['price']

# Add a constant for intercept
X = sm.add_constant(X) 

# Fit the model
model = sm.OLS(y, X).fit()

# Display summary
print(model.summary());

# results
print("Decision: Reject' the null hypothesis at alpha 0.05.")
print("Conclusion: There is sufficient evidence to claim The number of bedrooms  has an effect on price.")
```

## Objective 4: Is there a significant price difference between furnished and unfurnished houses?

Since furnishingstatus has three categories (furnished, semi-furnished, unfurnished), 

I will  use ANOVA (Analysis of Variance).

(One-Way ANOVA Test)

```python
# Define hypothesis
null_hypothesis = "There is no significant price difference between furnishing statuses."
alternate_hypothesis = "There is a significant price difference between furnished, semi-furnished, and unfurnished houses"

# Extract price data for each furnishing category
furnished = df_encoded[df_encoded['furnishingstatus'] == 2]['price']
semi_furnished = df_encoded[df_encoded['furnishingstatus'] == 1]['price']
unfurnished = df_encoded[df_encoded['furnishingstatus'] == 0]['price']

# Perform ANOVA test
f_stat, p_value = stats.f_oneway(furnished, semi_furnished, unfurnished)

# Decision
alpha = 0.05
if p_value <= alpha:
    decision = "Reject"
    conclusion = "There is a significant difference in price among furnishing statuses."
else:
    decision = "Fail to reject"
    conclusion = "There is no significant difference in price among furnishing statuses."

print(f"F-statistic: {f_stat:.4f}, P-value: {p_value}")
print(f"Decision: {decision} the null hypothesis at alpha={alpha}.")
print("Conclusion:", conclusion)
```

## Objective 5: Does having air conditioning increase the price?


```python
# Define hypothesis
null_hypothesis = "Houses with air conditioning have the same average price as those without"
alternate_hypothesis = "Houses with air conditioning have a higher price."

# Perform t-test
t_stat, p_value = stats.ttest_ind(df_encoded['airconditioning'],df_encoded['price'])

# Decision
alpha = 0.05
if p_value <= alpha:
    decision = "Reject"
    conclusion = "There is sufficient evidence to suggest that air-conditioned houses have a different price."
else:
    decision = "Fail to reject"
    conclusion = "There is no significant evidence that air-conditioned houses have different prices."

print(f"T-statistic: {t_stat:.4f}, P-value: {p_value}")
print(f"Decision: {decision} the null hypothesis at alpha={alpha}.")
print("Conclusion:", conclusion)
```
