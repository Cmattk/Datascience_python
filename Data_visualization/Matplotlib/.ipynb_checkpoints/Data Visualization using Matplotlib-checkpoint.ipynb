# Data Visualization using Matplotlib in Python
Matplotlib is a powerful and widely-used Python library for creating static, animated and interactive data visualizations.
The library is built on top of NumPy making it efficient for handling large datasets.

Matplotlib provides a module called pyplot which offers a MATLAB-like interface for creating plots and charts. It simplifies the process of generating various types of visualizations by providing a collection of functions that handle common plotting tasks.

# Line plot styles

```python
import matplotlib.pyplot as plt
import random as random

students = ["Jane","Joe","Beck","Tom",
            "Sam","Eva","Samuel","Jack",
            "Dana","Ester","Carla","Steve",
            "Fallon","Liam","Culhane","Candance",
            "Ana","Mari","Steffi","Adam"]
marks = []
for i in range(0,len(students)):
    marks.append(random.randint(60,101))

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("CLASS RECORDS")
plt.plot(students, marks, 'g--o')
```

```python
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("CLASS RECORDS")
plt.plot(students, marks, color = 'green',
         linestyle = 'solid', marker = 'o',
         markerfacecolor = 'red', markersize = 12)
```

```python
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("CLASS RECORDS")
plt.plot(students, marks, 'm--o')

# Adding grid lines
plt.grid(True, which='both', linestyle='--', linewidth=0.5)

plt.show()
```

# Bar Plot
A bar plot uses rectangular bars to represent data categories, with bar length or height proportional to their values. It compares discrete categories, with one axis for categories and the other for values.

x: The categories (e.g., fruits).
height: The corresponding values (e.g., sales).
width: The width of the bars (default is 0.8).
bottom: The baseline for the bars (default is 0).
align: How to align bars (‘center’ or ‘edge’)

Syntax: plt.bar(x, height, width, bottom, align)

```python
import numpy as np
fruits = ['Apples', 'Bananas', 'Cherries', 'Dates']
sales = [400, 350, 300, 450]

plt.bar(fruits,sales,color = 'violet')
plt.title('Fruit Sales')
plt.xlabel('Fruits')
plt.ylabel('Sales')
plt.show()
```

```python
# horizontal barplot
plt.barh(fruits, sales, color = 'yellow')

plt.title('Fruit Sales')
plt.xlabel('Fruits')
plt.ylabel('Sales')
plt.show()
```

```python
# adjust width
plt.bar(fruits, sales, width=.4)
plt.title('Fruit Sales')
plt.xlabel('Fruits')
plt.ylabel('Sales')
plt.show()
```

# multiple bar plots
Multiple bar plots are used when comparison among the data set is to be done when one variable is changing. We can easily convert it as a stacked area bar chart, where each subgroup is displayed by one on top of the others. It can be plotted by varying the thickness and position of the bars. Following bar plot shows the number of students passed in the engineering branch:

```python
barWidth = 0.25
fig = plt.subplots(figsize = (12,8))

IT = [12, 30, 1, 8, 22] 
ECE = [28, 6, 16, 5, 10] 
CSE = [29, 3, 24, 25, 17] 

br1 = np.arange(len(IT))
br2 = [x + barWidth for x in br1]
br3 = [x + barWidth for x in br2]

plt.bar(br1, IT, color ='r', width = barWidth, 
        edgecolor ='grey', label ='IT') 
plt.bar(br2, ECE, color ='g', width = barWidth, 
        edgecolor ='grey', label ='ECE')
plt.bar(br3, CSE, color ='b', width = barWidth, 
        edgecolor ='grey', label ='CSE')

plt.xlabel('Branch', fontweight ='bold', fontsize = 15) 
plt.ylabel('Students passed', fontweight ='bold', fontsize = 15) 
plt.xticks([r + barWidth for r in range(len(IT))], 
        ['2015', '2016', '2017', '2018', '2019'])

plt.legend()
plt.show()
```

# Stacked bar plot
Stacked bar plots represent different groups on top of one another. The height of the bar depends on the resulting height of the combination of the results of the groups. It goes from the bottom to the value instead of going from zero to value.

bar plot represents the contribution of boys and girls in the team.

```python
N = 5

boys = (20, 35, 30, 35, 27)
girls = (25, 32, 34, 20, 25)
boyStd = (2, 3, 4, 1, 2)
girlStd = (3, 5, 2, 3, 3)
ind = np.arange(N)   
width = 0.35  

fig = plt.subplots(figsize =(10, 7))
p1 = plt.bar(ind, boys, width, yerr = boyStd)
p2 = plt.bar(ind, girls, width,
             bottom = boys, yerr = girlStd)

plt.ylabel('Contribution')
plt.title('Contribution by the teams')
plt.xticks(ind, ('T1', 'T2', 'T3', 'T4', 'T5'))
plt.yticks(np.arange(0, 81, 10))
plt.legend((p1[0], p2[0]), ('boys', 'girls'))

plt.show()
```

# histogram
hist() function in Python is used to create histograms, which are graphical representations of data distribution. It divides the data into bins (non-overlapping intervals) and counts the frequency of values in each bin, plotting them as bars.

Syntax: matplotlib.pyplot.hist(x, bins=None, range=None, density=False, histtype=’bar’, color=None, label=None)


x: The data to be represented in the histogram.
bins: Specifies the number of bins or the bin edges for the histogram.
range: The lower and upper range of the bins.
density: If True, the histogram is normalized to form a probability density.
histtype: Defines the type of histogram (e.g., ‘bar’ for a traditional bar histogram).
color: Sets the color of the bars.
label: Label for the histogram, used in legends.

```python
data = [32, 96, 45, 67, 76, 28, 79, 62, 43, 81, 70,61, 95, 44, 60, 69, 71, 23
       ,69, 54, 76, 67,82, 97, 26, 34, 18, 16, 59, 88, 29, 30, 66,23, 65, 72, 
       20, 78, 49, 73, 62, 87, 37, 68,81, 80, 77, 92, 81, 52, 43, 68, 71, 86]
plt.hist(data)
plt.show()
```

```python
from scipy.stats import norm
# Define mean and standard deviation
mu, sigma = 121, 21

# Generate 1000 random samples
x = np.random.normal(mu, sigma, 1000)

# Plot histogram
num_bins = 50  # Adjusted for clarity
n, bins, _ = plt.hist(x, num_bins, density=True, color='green', alpha=0.6, edgecolor='black')

# Overlay normal distribution curve
pdf = norm.pdf(bins, mu, sigma)
plt.plot(bins, pdf, 'r--', linewidth=2, label="Normal Distribution")

# Add mean and std deviation lines
plt.axvline(mu, color='blue', linestyle='dashed', linewidth=2, label='Mean (μ)')
plt.axvline(mu - sigma, color='purple', linestyle='dashed', linewidth=1, label='μ - σ')
plt.axvline(mu + sigma, color='purple', linestyle='dashed', linewidth=1, label='μ + σ')

# Labels, title, and legend
plt.xlabel('X-Axis',fontweight = 'bold')
plt.ylabel('Density',fontweight = 'bold')
plt.title('Histogram with Normal Distribution Curve', fontweight = 'bold')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Show plot
plt.show()
```

```python
# Visualize the Specific Bars of Histogram
# Generate a 10000 x 3 matrix of standard normal random variables
x = np.random.randn(10000, 3)
colors = ['green', 'blue', 'lime']
plt.hist(x, bins=20, density=True, histtype='bar', color=colors,label=colors)
plt.legend(fontsize=10)
plt.title('matplotlib.pyplot.hist() Example', fontweight='bold')

plt.show()
```

```python
# Stacked Histogram with Three Datasets


# Generate three different normal distributions
mu, sigma = 121, 21
x1 = np.random.normal(mu, sigma, 1000)
x2 = np.random.normal(mu+10, sigma, 1000)
x3 = np.random.normal(mu-10, sigma, 1000)

# Colors for the datasets
colors = ['green', 'blue', 'lime']

# Plot histogram for three datasets
plt.hist([x1, x2, x3], bins=20, density=True, histtype='bar', color=colors, label=['Dataset 1', 'Dataset 2', 'Dataset 3'])

# Labels and Title
plt.legend(fontsize=10)
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Stacked Histogram Example', fontweight='bold')

# Show the plot
plt.show()
```

```python
# Assign Different Colors per Bin
# Generate data
mu, sigma = 121, 21
x = np.random.normal(mu, sigma, 1000)

# Create histogram with different bin colors
n, bins, patches = plt.hist(x, bins=20, density=True, histtype='bar', alpha=0.7)

# Assign different colors to each bar
for i, patch in enumerate(patches):
    patch.set_facecolor(plt.cm.viridis(i / len(patches)))  # Change colormap as needed

# Labels and Title
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Histogram with Gradient Colors', fontweight='bold')

# Show the plot
plt.show()
```

# Box Plot in Python
A Box Plot is also known as Whisker plot is created to display the summary of the set of data values having properties like minimum, first quartile, median, third quartile and maximum. In the box plot, a box is created from the first quartile to the third quartile, a vertical line is also there which goes through the box at the median. Here x-axis denotes the data to be plotted while the y-axis shows the frequency distribution.

Syntax:  matplotlib.pyplot.boxplot(data)
The notch = True attribute creates the notch format to the box plot, 
patch_artist = True fills the boxplot with colors, we can set different colors to different boxes.
The vert = 0 attribute creates horizontal box plot. 
labels takes same dimensions as the number data sets.

Median (central line in the box)
Interquartile Range (IQR) (box itself)
Whiskers (spread of data, typically 1.5 * IQR)
Outliers (individual points beyond whiskers)



```python
# Creating dataset
np.random.seed(10) # every time you run your code, you'll get the same sequence of random values.
# Generate data (Mean = 100, Std Dev = 20, Sample Size = 200)
data = np.random.normal(100, 20, 200)

fig = plt.figure(figsize =(10, 7))
# Creating plot
plt.boxplot(data, vert=True, patch_artist=True , boxprops=dict(facecolor="skyblue") )

# Labels and title
plt.ylabel("Value")
plt.title("Box Plot of Normally Distributed Data")

# show plot
plt.show()
```

```python
# Set seed for reproducibility
np.random.seed(10)

# Generate data
data_1 = np.random.normal(100, 10, 200)
data_2 = np.random.normal(90, 20, 200)
data_3 = np.random.normal(80, 30, 200)
data_4 = np.random.normal(70, 40, 200)
data = [data_1, data_2, data_3, data_4]

# Create figure
fig = plt.figure(figsize=(10, 7))
ax = fig.add_axes([0, 0, 1, 1])

# Box plot with custom colors
box_colors = ['skyblue', 'lightgreen', 'lightcoral', 'gold']

# Create box plot
bp = ax.boxplot(data, patch_artist=True)

# Apply colors to each box
for patch, color in zip(bp['boxes'], box_colors):
    patch.set_facecolor(color)
"""
1. bp['boxes']:
This retrieves a list of matplotlib.patches.PathPatch objects, where each object represents a box in the box plot.
These boxes define the interquartile range (IQR) (from Q1 to Q3).

2. zip(bp['boxes'], box_colors):
This pairs each box (from bp['boxes']) with a color from box_colors.
If there are 4 boxes, it pairs them with the 4 colors in box_colors.

3. Looping Through Boxes:
for patch, color in zip(bp['boxes'], box_colors):
Iterates over each patch (box) and its corresponding color.

4. 
patch.set_facecolor(color):

Changes the fill color of the box to the corresponding color.
"""
# Add labels
ax.set_xticklabels(['Dataset 1 (μ=100, σ=10)', 'Dataset 2 (μ=90, σ=20)', 
                    'Dataset 3 (μ=80, σ=30)', 'Dataset 4 (μ=70, σ=40)'])

# Titles and labels
ax.set_title("Box Plot of Four Different Datasets", fontweight='bold')
ax.set_ylabel("Value")

# Add grid for better readability
ax.yaxis.grid(True, linestyle="--", alpha=0.7)

# Show plot
plt.show()
```

```python
fig = plt.figure(figsize =(10, 7))
ax = fig.add_subplot(111)

# Creating axes instance
bp = ax.boxplot(data, patch_artist = True,
                notch ='True', vert = 0)

colors = ['#0000FF', '#00FF00', 
          '#FFFF00', '#FF00FF']

for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)

# changing color and linewidth of
# whiskers
for whisker in bp['whiskers']:
    whisker.set(color ='#8B008B',
                linewidth = 1.5,
                linestyle =":")

# changing color and linewidth of
# caps
for cap in bp['caps']:
    cap.set(color ='#8B008B',
            linewidth = 2)

# changing color and linewidth of
# medians
for median in bp['medians']:
    median.set(color ='red',
               linewidth = 3)

# changing style of fliers
for flier in bp['fliers']:
    flier.set(marker ='D',
              color ='#e7298a',
              alpha = 0.5)
    
# x-axis labels
ax.set_yticklabels(['data_1', 'data_2', 
                    'data_3', 'data_4'])

# Adding title 
plt.title("Customized box plot")

# Removing top axes and right axes
# ticks
ax.get_xaxis().tick_bottom()
ax.get_yaxis().tick_left()
    
# show plot
plt.show()
```

# Matplotlib Scatter
matplotlib.pyplot.scatter() is used to create scatter plots, which are essential for visualizing relationships between numerical variables. Scatter plots help illustrate how changes in one variable can influence another, making them invaluable for data analysis.

```python
x = np.array([12, 45, 7, 32, 89, 54, 23, 67, 14, 91])
y = np.array([99, 31, 72, 56, 19, 88, 43, 61, 35, 77])

plt.scatter(x, y)
plt.show()
```
